```python
wdi = pd.read_csv('WorldBankData.csv')
```

```python
year_cols = []

for col in wdi.columns:
    if col.startswith("19") or col.startswith("20"):
        year_cols.append(col)

year_cols
```

```python
wdi
```

```python
wdi_long = wdi.melt(
    id_vars=['Country Name', 'Country Code', 'Series Name'],
    value_vars=year_cols,
    var_name='year',
    value_name='value'
)

wdi_long
```

```python
wdi_long['year'] = wdi_long['year'].str[:4].astype(int)
wdi_long
```

```python
wdi_long.rename(columns={
    "Country Name": "country_name",
    "Country Code": "country_text_id"}, inplace=True)
```

```python
wdi_long['value'] = pd.to_numeric(wdi_long['value'], errors='coerce')
```

```python
wdi_long
```

```python
wdi_wide = wdi_long.pivot_table(
    index=['country_name', 'year'],
    columns='Series Name',
    values='value'
).reset_index()
```

```python
wdi_wide
```

```python
wdi_wide['country_name'].unique()
```

```python
regime_labels = {0: 'Closed Autocracy', 1: 'Electoral Autocracy', 
                 2: 'Electoral Democracy', 3: 'Liberal Democracy'}

latest['regime_label'] = latest['v2x_regime'].map(regime_labels)
```

```python
latest['year_label'] = latest['year'].astype(str)
```

```python
cv_vt_scatter = (
    ggplot(latest, aes(x='v2x_cspart',y='vap_turnout'))
    + geom_point(aes(color='regime_label'),
                 tooltips = layer_tooltips()
                 .line('@country_name')
                 .line('Turnout: @vap_turnout')
                 .line('Civic Participation: @v2x_cspart')
                 .line('Year: @year_label'),
                 size=6)
    + labs(
        title = 'Relationship between Civil Society Participation Index & Voter Turnout',
        subtitle = 'Higher turnout does not mean deeper civic engagement',
        x = 'Civil Society Participation Index (year of most recent election)',
        y = 'Voter Turnout (most recent election)',
        caption='Sources: V-Dem v15, IDEA Voter Turnout Database. \n Values shown for each country\'s most recent parliamentary election year.')
    + theme_light()
    + scale_y_continuous(limits = (0,100))
    + scale_x_continuous(limits = (0,1))
    + theme(plot_margin=[40, 20, 20, 20],
            text=element_text(family='Avenir Next'),
            plot_caption=element_text(size=8, color='grey', margin=[20,0,0,0]),
            legend_title=element_text(size=12, face='bold'),
            legend_text=element_text(size=11),
            plot_title=element_text(size=16, face='bold', margin=[0,0,10,0]),
            plot_subtitle=element_text(size=12, face='italic', margin=[0,0,10,0]),
            axis_title=element_text(size=12, face='italic'),
            plot_background=element_rect(fill='white'),
            panel_background=element_rect(fill='white'),
            panel_grid_major=element_line(color='#EBEBEB', size=0.5))
    + ggsize(800, 500)
    + scale_color_manual(
        name = 'Regime Type',
        values = {
            'Liberal Democracy': '#457B9D',
            'Electoral Democracy': '#2A9D8F',
            'Electoral Autocracy': '#E9C46A',
            'Closed Autocracy': '#E63946'
        }
    )
)
cv_vt_scatter
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
latest[['turnout_z', 'cspart_z']] = scaler.fit_transform(
    latest[['vap_turnout', 'v2x_cspart']]
)
```

```python
latest
```

```python
latest_dumbbell = latest.copy()
```

```python
latest_dumbbell['gap'] = latest_dumbbell['turnout_z'] - latest_dumbbell['cspart_z']
```

```python
latest_dumbbell = latest_dumbbell.sort_values(by='gap')
latest_dumbbell['country_name'] = pd.Categorical(
    latest_dumbbell['country_name'],
    categories=latest_dumbbell['country_name'],
    ordered=True)
```

```python
latest_dumbbell['turnout_label'] = 'Voter Turnout'
latest_dumbbell['cspart_label'] = 'Civic Participation'
```

```python
cv_vt_dumbbell = (
    ggplot(latest_dumbbell, aes(y='country_name'))
    + geom_point(aes(x='turnout_z', color='turnout_label'), size = 5,
                 tooltips = layer_tooltips()
                 .line('Turnout: @vap_turnout')
                 .line('Year: @year_label'))
    + geom_point(aes(x='cspart_z', color='cspart_label'), size = 5,
                 tooltips = layer_tooltips()
                 .line('Civic Participation: @v2x_cspart')
                 .line('Year: @year_label'))
    + geom_segment(aes(x='turnout_z', xend='cspart_z', y='country_name', yend='country_name'),
                   color='grey', alpha=0.5, size=1)
    + labs(title = 'Gap between Voter Turnout and Civic Participation',
           x = 'Voter Turnout and Civil Society Participation (standardised)',
           y = 'Country',
           caption='Sources: V-Dem v15, IDEA Voter Turnout Database. \n Values shown for each country\'s most recent parliamentary election year. ' \
           '\n Civic Participation Index and Voter Turnout have been standardised to the same scale for comparison')
    + ggsize(700, 700)
    + theme_light()
    + theme(plot_margin=[20, 10, 10, 10], 
        text=element_text(family='Avenir Next'),
        plot_caption=element_text(size=8, color='grey', margin=[20,0,0,0]),
        legend_title=element_text(size=12, face='bold'),
        legend_text=element_text(size=11),
        plot_title=element_text(size=16, face='bold', margin=[0,0,10,0]),
        plot_subtitle=element_text(size=12, face='italic', margin=[0,0,10,0]),
        axis_title=element_text(size=12, face='italic'),
        axis_text=element_text(size=10),
        panel_grid_major_x=element_line(color='#EBEBEB', size=0.3),
        panel_grid_major_y=element_blank(),
        panel_grid_minor=element_blank(),
        panel_border=element_blank())
    + scale_color_manual(
        name = 'Legend',
        values = {
            'Voter Turnout': '#264653',
            'Civic Participation': '#F4A261'
        }
    ))

cv_vt_dumbbell
```
